```python
#import torch library and neural network module
import torch 
import torch.nn as nn
```

```python
#Create a inputs as tensor(random input)
inputs = torch.tensor([3.0,1.0,-2.0])


```

Assign inputs to the neuron

```python
#Create a neuron
neuron = nn.Linear(in_features=3, out_features=1)
output = neuron(inputs)
```

```python
print(output)
```

```python
print(neuron.weight)
print(neuron.bias)
```

```python
neuron.weight @ inputs.T + neuron.bias
```
