```python
def haversine_distance(lat1,long1,lat2,long2):
    lat1 = (np.pi/180)*lat1 
    long1 = (np.pi/180)*long1 
    lat2 = (np.pi/180)*lat2
    long2 = (np.pi/180)*long2
    delta_Lat = lat2 - lat1
    delta_long = long2 - long1
    
    a1 = np.sin(delta_Lat/2)
    a1 = a1**2
    a2 = np.cos(lat1) * np.cos(lat2)
    a3 = np.sin(delta_long/2)
    a3 = a3**2
    
    a = a1 + a2 * a3
    p1 = np.sqrt(a)
    p2 = np.sqrt(1-a)
    c = 2 * np.arctan2(p1,p2)

    d = 6371 *c
    return d
     
```

```python
trip_distance = haversine_distance(df['pickup_latitude'],df['pickup_longitude'],df['dropoff_latitude'],df['dropoff_longitude'])
trip_distance
```

```python
from sklearn.tree import DecisionTreeRegressor
min_error = 1000
dict_error = {}
for i in range (5,30):
    decisionTree = DecisionTreeRegressor(max_depth=i)
    decisionTree.fit(X_train_scalled,y_train)
    y_pred = decisionTree.predict(X_test_scalled)
   
    error = mean_absolute_error(y_pred, y_test)
    min_error = min(min_error,error)
    dict_error[i] = error

best_depth = min(dict_error, key = dict_error.get)




```
